# 01. Exploratory Data Analysis (EDA)
### Research Study: Medical Health Insurance Cost Prediction
This notebook inspects the primary 986-record clinical dataset (Orji & Ukwandu 2024).

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv('../data/raw/Medicalpremium.csv')
print('Dataset Shape:', df.shape)
df.head()
```

```python
# Missing values and data types
print(df.info())
print('Missing Values:\n', df.isnull().sum())
print('Duplicates:', df.duplicated().sum())
```

```python
# Distribution of Target Variable
plt.figure(figsize=(9, 4.5))
sns.histplot(df['PremiumPrice'], kde=True, color='#2563EB', bins=30)
plt.title('Distribution of Annual Premium Price', weight='bold')
plt.xlabel('Premium Price')
plt.show()
```

```python
# Correlation Heatmap
plt.figure(figsize=(10, 7))
sns.heatmap(df.corr(), annot=True, fmt='.2f', cmap='Blues', linewidths=0.5)
plt.title('Feature Correlation Matrix', weight='bold')
plt.show()
```
